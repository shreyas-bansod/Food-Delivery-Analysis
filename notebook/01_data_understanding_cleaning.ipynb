## Missing Value Treatment

After converting placeholder values such as `"NaN "` into actual missing values, four columns contain missing data.

| Column | Missing % | Treatment |
|---|---:|---|
| Delivery_person_Ratings | 4.30% | Kept as NaN |
| Delivery_person_Age | 4.15% | Kept as NaN |
| Time_Orderd | 3.80% | Kept as NaT |
| Weatherconditions | 1.35% | Kept as NaN |

Missing values were not replaced with arbitrary values such as 0 because doing so could distort the analysis. Rows will only be excluded from specific analyses when the required variable is missing.

## Data Cleaning Summary

The dataset was cleaned using a reproducible process while preserving the original raw data.

### Cleaning actions performed

- Converted placeholder `"NaN "` values into proper missing values.
- Converted numeric columns such as delivery partner age and ratings to numeric data types.
- Removed the `"(min) "` text from `Time_taken(min)` and converted it to integer.
- Converted `Order_Date` to datetime.
- Converted `Time_Orderd` and `Time_Order_picked` to time-based datetime values.
- Identified and handled invalid delivery partner ratings above the expected range.
- Identified and handled invalid delivery partner ages below the expected working-age threshold.
- Converted invalid restaurant coordinates represented by `(0, 0)` to missing values.
- Preserved rows with missing values instead of deleting them unnecessarily.
- Verified that the dataset contains no duplicate rows.
- Confirmed that the target variable `Time_taken(min)` contains no missing values.

### Final Dataset

- Rows: 45,593
- Columns: 20
- Duplicate rows: 0
- Target missing values: 0

The cleaned dataset was saved as `food_delivery_cleaned.csv`.
